# Integrating TurtleBot 4 Mobility into the Testbed Controller (TC)

## Overview

This document explains how to integrate mobility into our Testbed Controller (TC) architecture, after we set it up in the previous chapter [link]. This extension adds a framework for a dynamic testbed, where RPi nodes can run tests with a quasi-static movement profile (e.g., forward, rotate, etc.) for data collection (e.g., iperf, RSSI, raw I/Q).

The base architecture consists of a TurtleBot 4 (TB4), developed by Clearpath Robotics, an Ethernet cable for a simple network connection, a USB-Ethernet adapter, RPi nodes for overhead communication, and access points (APs). Because this is the latest extension developed by the lab, we assume you have read the previous chapters and understand the core idea of the testbed and its setup:

* Testbed Architecture (Chap. 0)
* Setting Up the TC (Chap. 1)
* Setting Up RPi Nodes (Chap. 2)
* Control Network Configuration (Chap. 3)
* Test Network Configuration (Chap. 4)

Further inquire questions, see our Q&A section [] for debugging.

## Single-Node TB4 with Overhead RPi Controller

![Figure 1: Physical setup](./Images/section1_setup.png)

The image above, shown a basic single-node tb4 with overhead RPi Controller, after follow the installation process in the previous chapter. The Overhead Pi now have a **username: ucanlab**, **IP:10.1.1.X**, on the other hand the TB4's Pi have a default **username: ubuntu**, **IP:192.168.186.3**. Knowing the IP we can use the usb-ethernet adapter, plug in the ethernet cable and connected USB-B port from Overhead's Pi to the TB4's Pi port (i.e., as shown as image above). Afterward open up the Overhead's Pi desktop and configure the network as such (i.e., referenced to chap. 2). 

* `Setting` $ \rightarrow$ `Networks` $\rightarrow$ `USB Ethernet` (i.e., this will pop up after USB-Adapter inserted) $\rightarrow$ `IPv4` (i.e., set Address: `192.168.185.2`; Netmask: `255.255.255`) $\rightarrow$ `Disabled IPv6` $\rightarrow$ `Apply & Saved`  

Since our TB4's pi have an IP of 192.168.185.3, we want a similar subnet for the USB-Adapter via eth1 connection, thus an established IP reserved for the adpater at 192.168.185.2, we let's us communicate via ssh from overhead Pi to the TB4's Pi. 

If everything is configured as instructed, in the terminal the Overhead's Pi should be able to. 

- 1st: SSH inside the TB4's Pi.
    * `ssh ubuntu@192.168.185.3` (i.e., there'll be a password requires depend on your setup) 

- 2nd: activate ROS2/Jazzy and get related ROS2 packages.
    * `source opt/ros/jazzy/setup.bash` $\rightarrow$ `ros2 topic list | grep tb_01`

* Since this documents also included some simple movements scripts already developed, a 3rd check can also be running the `move_forward.py` and `rotate.py` scripts via the Overhead's ubuntu terminal.

    * `python3 move_forward.py` or `python3 rotate.py`

## Single-Node TB4 with Testbed Controller (TC)

![Figure 2: Physical setup](./Images/section2_setup.png)


This was an extension build upon what we already have in the previous section, now instead of control the tb4's via the overhead's Pi what we would like is to linked our Testbed Controller (TC) ontop of the overhead Pi for the purposes of scaleable/repeatable/automated as our lab intended. And to do this we need to linked the overhead Pi in our control network (i.e., using DHCP, visit chap. 3). To linked the Overhead Pi inside the control network (i.e., ucanlab_control), we need to reserve the IP with their MAC address, thus to identify the MAC address there're two ways for us to do it, first follow the instructions on chap. 4 or second just type this command in the Overhead's Pi terminal `cat /sys/class/net/wlan0/address`, there will be a serial MAC number appear for the Pi. Put this MAC command and IP associated from its overhead Pi, and click apply (for more details, look at the image below).

The TC have a default **IP:10.1.1.1**, already hooked up to the control network (visit chap.  1), to verify that the TC able to ping the overhead Pi use `bash verify_nodes.sh`, the status will display up if everything set correctly. Now from the TC, we need to enable the passwordless SSH (visit chap. 3) to do this in the TC type (i.e., this assumed for the key already generated in TC).

* `ssh-copy-id -i ~/.ssh/id_ed25519 ucanlab@10.1.1.X`

After we enabled the passwordless ssh, the process should be the same by `ssh ucanlab@10.1.1.X` activate the ROS2/Jazzy `source opt/ros/jazzy/setup.bash` and run `python3 move_forward.py`, since we ssh over the Pi everything should be the same.
