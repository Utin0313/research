# Integrating TurtleBot 4 Mobility into the Testbed Controller (TC)

## Overview

This document explains how to integrate mobility into our Testbed Controller (TC) architecture, after we set it up in the previous chapter [link]. This extension adds a framework for a dynamic testbed, where RPi nodes can run tests with a quasi-static movement profile (e.g., forward, rotate, etc.) for data collection (e.g., iperf, RSSI, raw I/Q).
The base architecture consists of a TurtleBot 4 (TB4), developed by Clearpath Robotics, an Ethernet cable for a simple network connection, a USB-Ethernet adapter, RPi nodes for overhead communication, and access points (APs). Because this is the latest extension developed by the lab, we assume you have read the previous chapters and understand the core idea of the testbed and its setup:

* Testbed Architecture (Chap. 0)
* Setting Up the TC (Chap. 1)
* Setting Up RPi Nodes (Chap. 2)
* Control Network Configuration (Chap. 3)
* Test Network Configuration (Chap. 4)

## Single-Node TB4 with Overhead RPi Controller

![Figure 1: Physical setup](./Images/section1_setup.png)

The image above, shown a basic single-node tb4 with overhead RPi Controller, after follow the installation process in the previous section. The Overhead Pi now have a **username: ucanlab**, **IP:10.1.1.X**, on the other hand the TB4's Pi have a default **username: ubuntu**, **IP:192.168.186.3**. Knowing the IP we can use the usb-ethernet adapter, plug in the ethernet cable and connected USB-B port from Overhead's Pi to the TB4's Pi port (i.e., as shown as image above). Afterward open up the Overhead's Pi desktop and configure the network as such (i.e., referenced to chap. 2). 

* `Setting` $ \rightarrow$ `Networks` $\rightarrow$ `USB Ethernet` (i.e., this will pop up after USB-Adapter inserted) $\rightarrow$ `IPv4` (i.e., set Address: `192.168.185.2`; Netmask: `255.255.255`) $\rightarrow$ `Disabled IPv6` $\rightarrow$ `Apply & Saved`  

Since our TB4's pi have an IP of 192.168.185.3, we want a similar subnet for the USB-Adapter via eth0 connection, thus an established IP reserved for the adpater at 192.168.185.2, we let's us communicate via ssh from overhead Pi to the TB4's Pi. 

If everything is configured as instructed, in the terminal the Overhead's Pi should be able to. 

- 1st: SSH inside the TB4's Pi.
    * `ssh ubuntu@192.168.185.3` (i.e., there'll be a password requires depend on your setup) 

- 2nd: activate ROS2/Jazzy and get related ROS2 packages.
    * `source opt/ros/jazzy/setup.bash` $\rightarrow$ `ros2 topic list | grep tb_01`

* Since this documents also included some simple movements scripts already developed, a 3rd check can also be running the `move_forward.py` and `rotate.py` scripts via the terminal.

    * `ssh ubuntu@192.168.185.3` $\rightarrow$ `python3 move_forward.py` or `python3 rotate.py`
