# Integrating TurtleBot 4 Mobility into the Testbed

[Img]

## Overview

This document explains how to integrate mobility into our Testbed Controller (TC) architecture. This extension adds a framework for a dynamic testbed, where RPi nodes can run tests with a quasi-static movement profile (e.g., forward, rotate, etc.) for data collection (e.g., iperf, RSSI, raw I/Q).
The base architecture consists of a TurtleBot 4 (TB4), developed by Clearpath Robotics; an Ethernet cable for a simple network connection; RPi nodes for overhead communication; and access points (APs).  Because this is the latest extension developed by the lab, we assume you have read the previous chapters and understand the core idea of the testbed and its setup:

* Testbed Architecture (Chap. 0)
* Setting Up the TC (Chap. 1)
* Setting Up RPi Nodes (Chap. 2)
* Control Network Configuration (Chap. 3)
* Test Network Configuration (Chap. 4)

We also assume your TB4 and its Create 3 packages are already fully set up per [section/link]. By splitting the instructions into the steps below, this document aims to cover the network configuration needed for **remote multi-node movement**, which is our primary objective for the Ubiquitous Communications and Networking (UCaN) Lab.

## Single-Node TB4 with Overhead RPi Controller

An overview of our physical setup can be seen in the image below.

![Figure 1: Physical setup](./Images/section1_setup.png)

In this setup, we have an overhead Pi with the default **username: ucanlab** and an IP on the typical lab subnet, **IP: 10.1.1.X** (as explained in Chap. 2). Inside the TB4, there is another Pi that has already been configured with Create 3 and ROS 2 Jazzy for control (as explained in [section/link]). For the purpose of keeping everything consistent with our lab work methodology, we keep the TB4 defaults: **username: ubuntu**, **IP: 192.168.185.3**. The monitor is connected via a micro-HDMI cable, and the keyboard and mouse are connected directly via USB-A cables to the USB 2.0 ports on the overhead Pi. The purpose is to double-check the connection with the TB4's Pi and to install the related software, which we will discuss below.


This first section shows the instructions to install Robot Operating System 2 (ROS 2) Jazzy on the overhead Pi, because our TurtleBot 4's internal Pi mainly uses ROS 2 Jazzy on **Ubuntu Server 24.04**. The overhead Pi also needs to run the same ROS 2 versio and operating system since the ROS2 come with rclpy which is essential for us to communicate via Data Distribution Serivce (DDS); thus, the overhead Pi running on top of the TB4 should ideally be a **Raspberry Pi 4 Model B** with **Ubuntu Desktop 24.04.X LTS (64-bit)** specifically (i.e., any version of Ubuntu Desktop 24.04 works, such as 24.04.2, 24.04.3, 24.04.4, etc., since they all contain the `noble` base repositories that ROS 2 Jazzy is designed for). If you have already followed the steps in Chap. 2, the idea is basically the same: we flash the micro-SD card using Raspberry Pi Imager and select Ubuntu Desktop 24.04 as the OS version.

![Figure 2: Physical setup](./Images/section1a_setup.png)
